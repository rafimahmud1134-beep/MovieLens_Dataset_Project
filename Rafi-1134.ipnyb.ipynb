```python


# Research Dataset Preparation
# Paper: Breaking the Curse of Quality Saturation
# Name: Rafi Mahmud
# ID: 0432320005101134
# Dataset: MovieLens-20M

# Step 1: Import Libraries

import numpy as np
import pandas as pd
import zipfile
import os
import urllib.request

from sklearn.model_selection import train_test_split

print("Name: Rafi Mahmud")
print("ID: 0432320005101134")


# Step 2: Download MovieLens-20M Dataset

url = "https://files.grouplens.org/datasets/movielens/ml-20m.zip"

urllib.request.urlretrieve(url, "ml-20m.zip")

with zipfile.ZipFile("ml-20m.zip", "r") as zip_ref:
    zip_ref.extractall(".")

print("\nDataset Downloaded Successfully!")


# Step 3: Load Dataset

data = pd.read_csv("ml-20m/ratings.csv")

print("\nOriginal Dataset:")
print(data.head())

print("\nDataset Shape:", data.shape)

print("\nTotal Users:", data["userId"].nunique())
print("Total Movies:", data["movieId"].nunique())
print("Total Ratings:", len(data))


# Step 4: Convert Ratings into Positive and Negative

# Rating >= 4 means Positive (1)
# Rating < 4 means Negative (0)

data["label"] = np.where(data["rating"] >= 4, 1, 0)

print("\nPositive and Negative Interactions:")
print(data["label"].value_counts())


# Step 5: Sort Data by Timestamp

data = data.sort_values("timestamp")


# Step 6: Create Positive Interaction Dataset

positive_data = data[data["label"] == 1].copy()

print("\nPositive Interaction Shape:", positive_data.shape)


# Step 7: User-Centric History (UCR)

# Keep the latest 512 positive movies for each user

user_history = (
    positive_data
    .groupby("userId")
    .tail(512)
    .groupby("userId")["movieId"]
    .apply(list)
)

print("\nUser-Centric History Created!")
print("Total Users:", len(user_history))


# Step 8: Item-Centric History (ICR)

# Keep the latest 512 positive users for each movie

item_history = (
    positive_data
    .groupby("movieId")
    .tail(512)
    .groupby("movieId")["userId"]
    .apply(list)
)

print("\nItem-Centric History Created!")
print("Total Movies:", len(item_history))


# Step 9: Train-Test Split (4:1)

train_data, test_data = train_test_split(
    data,
    test_size=0.20,
    random_state=42,
    stratify=data["label"]
)

print("\nTraining Dataset:", train_data.shape)
print("Testing Dataset:", test_data.shape)


# Step 10: Display Final Results

print("\n========== FINAL RESULTS ==========")

print("Researcher: Rafi Mahmud")
print("Student ID: 0432320005101134")

print("Total Ratings:", len(data))
print("Total Users:", data["userId"].nunique())
print("Total Movies:", data["movieId"].nunique())

print("Positive Interactions:", len(positive_data))

print("User History:", len(user_history))
print("Item History:", len(item_history))

print("Training Samples:", len(train_data))
print("Testing Samples:", len(test_data))

print("\nSample User History:")
print(user_history.head())

print("\nSample Item History:")
print(item_history.head())


# Step 11: Save Processed Dataset

train_data.to_csv("train_dataset.csv", index=False)
test_data.to_csv("test_dataset.csv", index=False)

positive_data.to_csv("positive_interactions.csv", index=False)

print("\nAll Datasets Saved Successfully!")
```
